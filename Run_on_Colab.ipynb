```python
# @title 2. Start Server & Generate Public URL
# We use localtunnel to securely expose port 3000 to the public web
import subprocess
import time
import urllib.request

print("Starting Node.js Server...")
server_process = subprocess.Popen(["node", "server.js"], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
time.sleep(3) # Wait for server to boot

print("\n=========================================")
print("   Server is running! Generating URL...")
print("=========================================\n")

# Fetch the Colab instance's IP address (required as a password for localtunnel)
ip = urllib.request.urlopen('https://ipv4.icanhazip.com').read().decode('utf8').strip()
print(f"\033[93mIMPORTANT: When prompted for a password on the website, enter this IP address:\033[0m \033[1m{ip}\033[0m\n")

!npx localtunnel --port 3000
```
