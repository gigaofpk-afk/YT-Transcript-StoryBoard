# YouTube Transcript Checker - Google Drive Batch Processor
This notebook runs the backend locally in the Colab instance, mounts your Google Drive, and automatically processes a folder of screenshots (Input) and saves the master ZIP transcripts to an Output folder.

```
# 1. Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')
```

```
# 2. Setup Directories in Drive
import os
INPUT_DIR = '/content/drive/MyDrive/YouTube_Input'
OUTPUT_DIR = '/content/drive/MyDrive/YouTube_Output'

os.makedirs(INPUT_DIR, exist_ok=True)
os.makedirs(OUTPUT_DIR, exist_ok=True)
print(f"Please upload your screenshots to: {INPUT_DIR}")
```

```
# 3. Install Node.js Dependencies
# Note: Upload the entire 'youtube-checker' folder to your Google Drive first,
# or upload the zip and extract it here.
PROJECT_PATH = '/content/drive/MyDrive/youtube-checker' # Change this if your path is different
%cd $PROJECT_PATH
!npm install
```

```
# 4. Run the Batch Processor
import subprocess
import time

TARGET_LANG = 'original' # Use 'en', 'ur', 'es' for automatic translation

print('Starting Backend Server in background...')
server = subprocess.Popen(['node', 'server.js'], stdout=subprocess.PIPE, stderr=subprocess.PIPE)
time.sleep(3) # Wait for server to boot

print('Running Automated Screenshot Processor...')
!env INPUT_DIR="$INPUT_DIR" OUTPUT_DIR="$OUTPUT_DIR" TARGET_LANG="$TARGET_LANG" node process_folder.js

server.terminate()
print('✅ All screenshots processed!')
```
